<h1>Integrating the genre API inplace of the ids</h1>

```python
genre_res=requests.get("https://api.themoviedb.org/3/genre/movie/list?api_key=8265bd1679663a7ea12ac168da84d2e8&language=en-US")
```

```python
genre_res.json()
```

```python
genre_data=genre_res.json()
```

```python
genre_mapping = {
    genre['id']: genre['name']
    for genre in genre_data['genres']
}
```

```python
genre_mapping
```

```python
def convert_genres(genre_ids):
    return [genre_mapping[genre_id] for genre_id in genre_ids]
```

```python
df['genres'] = df['genre_ids'].apply(convert_genres)
```

```python
df = df[['adult', 'title', 'overview', 'popularity', 'release_date', 'genres']]
```

<h1>Starting Text Preprocessing </h1>

```python
df['genres'] = df['genres'].apply(lambda x: ' '.join(x))
```

```python
df.sample(2)
```

<h5>Lowercasing</h5>

```python
df[["title", "overview", "genres"]] = df[["title", "overview", "genres"]].apply(lambda x: x.str.lower())
```

```python
df.sample(5)
```

<h5>Removing Punctuation</h5>

```python
import string
exclude=string.punctuation.replace("'", "")

def remove_punc(text):
    return text.translate(str.maketrans(exclude," " * len(exclude)))
```

```python
df[["title", "overview"]] = df[["title", "overview"]].map(remove_punc)
```
