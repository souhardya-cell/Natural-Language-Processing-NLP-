<h2>Using the split function</h2>

```python
#word tokenization
sent1="I am going to Delhi"
sent1.split()
```

<h3>Problems with split function</h3>

```python
sent3 = 'I am going to delhi!'
sent3.split()
```

<p>The punctuation marks is getting separated along with delhi</p>

```python
sent4 = 'Where do you think I should go? I have 3 day holiday'
sent4.split('.')
```

<p>As the sentence here is separated with "?", our split function fails universally</p>

<h2>Using Regular Expression</h2>

```python
import re
sent3 = 'I am going to delhi!'
tokens = re.findall("[\w']+", sent3)
tokens
```

```python
text = """Lorem Ipsum is simply dummy text of the printing and typesetting industry? 
Lorem Ipsum has been the industry's standard dummy text ever since the 1500s, 
when an unknown printer took a galley of type and scrambled it to make a type specimen book."""
sentences = re.compile('[.!?] ').split(text)
sentences
```

<h2>Using NLTK</h2>

```python
import nltk
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize,sent_tokenize
```

```python
sent4 = 'I am going to visit delhi!'
word_tokenize(sent4)
```

<p>PROBLEM SOLVED</p>

```python
text = """Lorem Ipsum is simply dummy text of the printing and typesetting industry? 
Lorem Ipsum has been the industry's standard dummy text ever since the 1500s, 
when an unknown printer took a galley of type and scrambled it to make a type specimen book."""

sent_tokenize(text)
```
