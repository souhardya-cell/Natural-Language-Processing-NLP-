```python
!pip install gensim

import gensim

from gensim.models import Word2Vec,KeyedVectors
```

```python
model = KeyedVectors.load_word2vec_format(
    "GoogleNews-vectors-negative300.bin",
    binary=True,
    limit=500000
)
```

```python
model['cricket']
```

<h5>Each word gets a collection of 300 dimensional vectors</h5>

<h3>Finding most similar words to a word</h3>

```python
model.most_similar('man')
```

```python
model.most_similar('cricket')
```

```python
model.most_similar('ODI')
```

<h3>Finding the similarity between two words</h3>

```python
model.similarity('man','woman')
```

```python
model.similarity('man','PHP')
```

```python
#finding odd one out from a list
model.doesnt_match(['PHP','java','monkey'])
```

<h3>Applying vector maths</h3>

```python
vec=model['king']-model['man']+model['woman']
#king-man+woman= queen
model.most_similar([vec])
```

```python
vec=model['INR']-model['India']+model['England']
#INR-India+England= GBP
model.most_similar([vec])
```

<p><b>We worked with the pre-trained Google News Word2Vec model, which contains word representations learned from a very large news corpus, with each word represented as a 300-dimensional vector. From this, we understood that Word2Vec converts words into dense, low-dimensional vectors while capturing semantic relationships between words based on their context. Unlike BoW and TF-IDF, it can identify relationships such as King − Man + Woman ≈ Queen, find similar words using vector similarity, and allows vector arithmetic.</b></p>
