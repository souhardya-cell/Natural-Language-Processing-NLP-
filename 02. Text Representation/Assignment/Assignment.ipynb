<h1>Text Preprocessing</h1>

<h5>Lowercasing</h5>

```python
df['review']=df['review'].str.lower()
```

```python
import re
def remove_html(review):
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',review)
```

<h5>Removing HTML tags</h5>

```python
df['review']=df['review'].apply(remove_html)
```

```python
df['review'][0]
```

<h5>Handling Contractions</h5>

```python
import contractions
def expand_contractions(text):
    return contractions.fix(text)
```

```python
df['review']=df['review'].apply(expand_contractions)
```

<h5>Removing Punctuations</h5>

```python
import string
exclude=string.punctuation

def remove_punc(review):
    return review.translate(str.maketrans(exclude," " * len(exclude)))

#removing puncs and adding spaces in place of them.
```

```python
df['review']=df['review'].apply(remove_punc)
```

```python
df['review'][0]
```

<h5>Removing Extra Spaces</h5>

```python
def remove_extra_spaces(text):
    return " ".join(text.split())
```

```python
df['review']=df['review'].apply(remove_extra_spaces)
```

```python
df['review'][0]
```

<h5>Stopword Removal</h5>

```python
import nltk
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    new_text=[]
    for word in text.split():
        if word in stopwords.words("english"):
            new_text.append(" ")
        else:
            new_text.append(word)
    x=new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
df['review']=df['review'].apply(remove_stopwords)
```

```python
df['review'][0]
```

```python
df['review']=df['review'].apply(remove_extra_spaces)
```

<h5>Tokenization</h5>

```python
from nltk.tokenize import word_tokenize
```

```python
def tokenize(text):
    return word_tokenize(text)
```

```python
df['review']=df['review'].apply(tokenize)
```

```python
df['review'][0]
```

<h5>Lemmatization</h5>

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
def lemmatize_text(text):
    return " ".join(
        lemmatizer.lemmatize(word,pos="v")
        for word in text
    )
```

```python
df['review']=df['review'].apply(lemmatize_text)
```
