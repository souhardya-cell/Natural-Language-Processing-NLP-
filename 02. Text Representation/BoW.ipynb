```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    'text': [
        'people watch campusx',
        'campusx watch campusx',
        'people write comment',
        'campusx write comment'
    ],
    'output': [1, 1, 0, 0]
})
```

```python
df
```

<h2>Bag of Words</h2>

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()
```

```python
bow=cv.fit_transform(df['text'])
```

```python
#vocabulary
print(cv.vocabulary_)
```

```python
print(bow[0].toarray())
```

<p>people watch campusx
so as we can see the vocabulary index, campusx is 0, followed by comment,people,watch and write.

So for the first text, 
1,0,1,1,0 represents campusx, people and watch buffed</p>

```python
print(bow[1].toarray())
```

<h3>Solves the OOV problem to some extent</h3>

```python
cv.transform(["people watch campusx videos and comment on campusx videos"]).toarray()
```

<p>Here even though the input has new vocabulary which werent present in the original dataframe, BoW ignores them and returns the freq array on the basis of the vocabulary words it is familiar with. Thus, it solves the Out Of Vocabulary (OOV) problem to an extent.

<b>However, the problem arises that it is completely ignoring the newly added words in the input thus the whole meaning of the sentence is ignored.</b></p>
